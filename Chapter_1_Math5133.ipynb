A differential equation (DE) is an equation that relates a function with one or more of its derivatives. In other words, it describes how a quantity changes in relation to other quantities. Differential equations are fundamental tools in mathematics, physics, engineering, biology, economics, and many other fields because they model dynamic systems  that evolve over time or space.



### Why are Differential Equations Important?

Differential equations (DEs) are among the most powerful and widely used mathematical tools for modeling dynamic processes in the natural and engineered world. Their significance arises from their unique ability to encode the *rate of change* of a system as a function of its current state and external influences. This makes them indispensable across scientific and engineering disciplines for the following reasons:  

**I. Modeling of Real-World Phenomena:**  
    Differential equations naturally describe systems that evolve continuously over time or space. Examples include:
    
 - *Population dynamics*: The logistic or exponential growth models, $\frac{dP}{dt} = kP$, capture how populations change under resource constraints or ideal conditions.
- *Radioactive decay*: Governed by $\frac{dN}{dt} = -\lambda N$, where the decay rate is proportional to the number of undecayed nuclei.
- *Heat conduction*: Described by the heat equation $\frac{\partial u}{\partial t} = \alpha \nabla^2 u$, a partial differential equation modeling temperature diffusion in solids.
 - *Fluid flow*: The Navier--Stokes equations model velocity, pressure, and density fields in fluids.
- *Electrical circuits*: Kirchhoff’s laws lead to ODEs such as $L\frac{dI}{dt} + RI = V(t)$ for an RL circuit.
- *Celestial mechanics*: Newton’s law of gravitation yields second-order ODEs that predict planetary orbits.


**II. Predictive Capability:**  
    Given an initial (or boundary) condition, solving a differential equation provides the state of the system at any future (or past) time. This predictive power is foundational in simulation, control theory, and forecasting from climate models to epidemic spread (e.g., SIR models).


**III. Fundamental Laws of Physics Are Expressed as DEs:**  
    The core principles of classical and modern physics are inherently differential:
- *Newton’s second law*: $F = m\frac{d^2x}{dt^2}$ is a second-order ODE relating force to acceleration.
- *Maxwell’s equations*: A system of PDEs governing electric and magnetic fields, forming the foundation of electromagnetism.
- *Schrödinger’s equation*: $i\hbar \frac{\partial \psi}{\partial t} = \hat{H}\psi$, a PDE central to quantum mechanics.
- *Einstein’s field equations*: Nonlinear PDEs that describe gravity in general relativity.
    Thus, differential equations are not merely mathematical abstractions, they are the *language of physical law*.

Consequently, the study, analysis, and numerical solution of differential equations particularly through robust, efficient, and stable algorithms remains a central theme in applied mathematics and computational science.

## 1.1 What is an Initial Value Problem?

An **Initial Value Problem (IVP)** consists of a differential equation together with a specified value of the unknown function at a particular point — usually at the "start" of the process.

For a first-order ordinary differential equation (ODE), the general form of an IVP is:
$$
\left\{
\begin{aligned}
\frac{dy}{dt} &= f(t, y), \quad a \leq t \leq b,\\
y(a) &= \alpha,
\end{aligned}
\right.
\label{eq:ivp}
$$
where:  
$ y(t) $ is the unknown function (e.g., position, temperature, population),  
$ f(t, y) $ describes the rate of change of $ y $,  
$ y(a) = \alpha $ is the **initial condition**, representing the state of the system at time $ t = a $.

The goal is to find a function $ y(t) $ that satisfies both the differential equation and the initial condition.

 **Remark**  
 Unlike boundary value problems (BVPs), where conditions are imposed at multiple points, IVPs model evolution over time, starting from a known state.


### Physical Motivation: Why IVPs Matter?

Many natural processes evolve according to **laws of change**. These are often expressed as differential equations. The initial condition reflects our knowledge of the system at a specific moment  like "where it starts"  and allows us to predict its future behavior. We now explore several classical examples.


### Example 1: Population Growth

Suppose a population of bacteria grows at a rate proportional to its current size. Let $ P(t) $ be the population at time $ t $. Then:
$$
\frac{dP}{dt} = kP,
$$
where $ k > 0 $ is the growth rate.  
If we know the initial population $ P(0) = P_0 $, the IVP becomes:
$$
\left\{
\begin{aligned}
\frac{dP}{dt} &= kP, \\
P(0) &= P_0.
\end{aligned}
\right.
$$
This models **exponential growth**. The solution is:
\[
P(t) = P_0 e^{kt}.
\]
Without the initial condition $ P(0) = P_0 $, we cannot determine the constant of integration. The initial condition **calibrates the model** to real data.

 **Remark**  
 In more realistic models, growth may slow due to limited resources (logistic model), but the IVP structure remains central.


### Example 2: Radioactive Decay

Radioactive substances decay at a rate proportional to the amount present. Let $ A(t) $ be the amount of substance at time $ t $. Then:
$$
\frac{dA}{dt} = -kA, \quad k > 0.
$$
Given an initial amount $ A(0) = A_0 $, the IVP is:
$$
\left\{
\begin{aligned}
\frac{dA}{dt} &= -kA, \\
A(0) &= A_0.
\end{aligned}
\right.
$$
The solution is: $A(t) = A_0 e^{-kt}.$  
This describes **exponential decay**. The half-life $ T_{1/2} = \frac{\ln 2}{k} $ can be computed from $ k $, but the actual amount at any time depends on $ A_0 $.

 **Remark**  
 In experiments, $ A_0 $ is measured at $ t = 0 $; without it, we cannot predict how much material remains after, say, 10 years.


### Example 3: Newton's Law of Cooling

A hot object cools in a surrounding medium at a rate proportional to the temperature difference. Let $ T(t) $ be the object's temperature, and $ T_m $ the constant ambient temperature. Then:
$$
\frac{dT}{dt} = -k(T - T_m), \quad k > 0.
$$
If the initial temperature is $ T(0) = T_0 $, the IVP is:
$$
\left\{
\begin{aligned}
\frac{dT}{dt} &= -k(T - T_m), \\
T(0) &= T_0.
\end{aligned}
\right.
$$
The solution is: $T(t) = T_m + (T_0 - T_m)e^{-kt}.$  
This shows how the object's temperature approaches $ T_m $ over time. The initial condition $ T_0 $ determines the starting point of the cooling curve.

### Example 4: Motion of a Falling Object (with Air Resistance)

Consider a mass $ m $ falling under gravity with air resistance proportional to velocity $ v $. Newton's second law gives:
$$
m \frac{dv}{dt} = mg - \gamma v,
$$
where $ \gamma > 0 $ is the drag coefficient.  
Divide by $ m $:
$$
\frac{dv}{dt} = g - \frac{\gamma}{m} v.
$$
with initial velocity $ v(0) = v_0 $, this forms the IVP:
$$
\left\{
\begin{aligned}
\frac{dv}{dt} &= g - \frac{\gamma}{m} v, \\
v(0) &= v_0.
\end{aligned}
\right.
$$
The solution approaches a **terminal velocity** $ v_t = \frac{mg}{\gamma} $. 

The initial condition determines how quickly this limit is approached. Consequently, the study, analysis, and numerical solution of differential equations particularly through robust, efficient, and stable algorithms remains a central theme in applied mathematics and computational science.


### 1.1.2 Why Initial Conditions are Essential?
The differential equation describes **how** a system changes, but not where it starts. Without an initial condition:
- We get a **family of solutions** (general solution),
-   But no way to pick the one that matches reality.
For example, the general solution to $ y' = y $ is $ y(t) = Ce^t $. Only when we impose $ y(0) = 2 $ do we get $ y(t) = 2e^t $ a specific, predictive model.

Thus, the initial condition **anchors the mathematical model to physical observation**.

#### Summary
- An IVP combines a differential equation with an initial condition: $ y(a) = \alpha $.
- IVPs model **time-evolving systems** in physics, biology, and engineering.
- Physical examples include:

   - Population growth: $ P' = kP $
    - Radioactive decay: $ A' = -kA $
    - Newton's Law of Cooling: $ T' = -k(T - T_m) $
     - Motion with drag: $ v' = g - cv $
- The initial condition ensures the solution is **unique and physically meaningful**.
- Without it, we cannot make quantitative predictions.

IVPs are the foundation for understanding dynamic behavior in the natural world.

### 1.1.3 Computational Examples with Python

#### Example 1.1 
Write the IVP for a population growing at 3 percent per year, starting from 1000 individuals.

#### Solution  

A population growing at 3 percent per year implies exponential growth governed by  $\frac{dP}{dt} = kP, \quad \text{with } k = 0.03.$

Given the initial population $ P(0) = 1000 $, the **initial value problem (IVP)** is:

$$
\left\{
\begin{aligned}
\frac{dP}{dt} &= 0.03\,P, \\
P(0) &= 1000.
\end{aligned}
\right.$$

The analytical solution is:
$
P(t) = 1000\,e^{0.03t}.$

```python
import numpy as np
import matplotlib.pyplot as plt

# Exercise 1: Population Growth (3% per year, P0 = 1000)
k1 = 0.03
P0 = 1000
t1 = np.linspace(0, 20, 200)  # 20 years
P1 = P0 * np.exp(k1 * t1)

plt.figure(figsize=(6, 5))  # Approx. 500x400 pixels (adjust DPI if needed)
plt.plot(t1, P1, 'b-', linewidth=2)
plt.xlabel('Time (years)')
plt.ylabel('Population P(t)')
plt.title('Population Growth at 3%/year')
plt.grid(True)
plt.tight_layout()
plt.show()

# Print the formula
print(f'P(t) = {P0}*exp({k1:.2f}t)')
```

#### Example 1.2: Newton’s Law of Cooling  

A cup of coffee at $90^\circ\text{C}$ cools in a room at $20^\circ\text{C}$ with cooling constant $k = 0.1~\text{min}^{-1}$.   Write the initial value problem (IVP) and compute the temperature after 5 minutes, $T(5)$.

### Solution  

Newton’s Law of Cooling states: $$
\frac{dT}{dt} = -k\,(T - T_m),$$
where  
- $T(t)$: temperature of the object (°C),  
- $T_m = 20^\circ\text{C}$: ambient (room) temperature,  
- $k = 0.1~\text{min}^{-1}$: cooling rate constant,  
- $T(0) = 90^\circ\text{C}$: initial coffee temperature.

Thus, the **IVP** is:
$$
\left\{
\begin{aligned}
\frac{dT}{dt} &= -0.1\,(T - 20), \\
T(0) &= 90.
\end{aligned}
\right.
$$

The analytical solution is:
$$
T(t) = T_m + (T_0 - T_m)e^{-kt} = 20 + 70\,e^{-0.1t}.
$$

At $t = 5$ minutes:
$$
T(5) = 20 + 70\,e^{-0.5} \approx 20 + 70 \times 0.60653 \approx 62.46^\circ\text{C}.
$$

```python
import numpy as np
import matplotlib.pyplot as plt
# Parameters
k = 0.1          # per minute
Tm = 20          # ambient temp (°C)
T0 = 90          # initial temp (°C)
# Analytical solution function
def T(t):
    return Tm + (T0 - Tm) * np.exp(-k * t)

# Compute T(5)
t_eval = 5
T_5 = T(t_eval)
print(f"T({t_eval}) = {T_5:.2f} °C")
# Plot temperature over time (0 to 30 minutes)
t_vals = np.linspace(0, 30, 300)
T_vals = T(t_vals)

plt.figure(figsize=(7, 4.5))
plt.plot(t_vals, T_vals, 'r-', linewidth=2, label=r'$T(t) = 20 + 70e^{-0.1t}$')
plt.axhline(Tm, color='k', linestyle='--', alpha=0.6, label=r'Ambient $T_m = 20^\circ$C')
plt.scatter([t_eval], [T_5], color='blue', zorder=5,
            label=rf'$T(5) \approx {T_5:.1f}^\circ\mathrm{{C}}$')
plt.xlabel('Time $t$ (minutes)', fontsize=12)
plt.ylabel('Temperature $T(t)$ (°C)', fontsize=12)
plt.title("Newton's Law of Cooling: Coffee Cooling", fontsize=13)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.xlim(0, 30)
plt.ylim(15, 95)
plt.tight_layout()
plt.show()
```

#### Example 1.3:  
Solving $ y' = 2y,\ y(0) = 3 $

Solve the initial value problem and interpret it as a population model.

#### Solution  

The differential equation  
$$
\frac{dy}{dt} = 2y
$$ 
is separable (and linear). Its general solution is  $
y(t) = Ce^{2t}.$

Applying the initial condition $ y(0) = 3 $:  
$$
3 = Ce^{0} = C \quad \Rightarrow \quad C = 3.
$$

Exact solution:  $y(t) = 3\,e^{2t}$


####  Interpretation  
- Models a population starting with **3 individuals**.  
- Growth rate is $ k = 2 $, i.e., **200% per unit time** (continuous compounding).  
- **Doubling time**:  
  $$
  T_{\text{double}} = \frac{\ln 2}{k} = \frac{\ln 2}{2} \approx 0.347 \text{ time units}.
  $$
  So the population doubles roughly every 0.35 units (e.g., hours, days  depending on context).

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
k = 2.0        # growth rate
y0 = 3         # initial population

# Time (short interval due to rapid growth)
t = np.linspace(0, 2, 200)
y = y0 * np.exp(k*t)

# Doubling time
T_double = np.log(2)/k

# Plot
plt.figure(figsize=(7, 4.5))
plt.plot(t, y, 'g-', linewidth=2.5, label=rf'$y(t) = {y0}e^{{{k}t}}$')
plt.axvline(T_double, color='gray', linestyle='--', alpha=0.7,
            label=rf'Doubling time $T_d \approx {T_double:.3f}$')
plt.scatter([0, T_double], [y0, 2*y0], color='red', zorder=5)

plt.xlabel('Time $t$ (units)', fontsize=12)
plt.ylabel('Population $y(t)$', fontsize=12)
plt.title('Exponential Growth: $y\' = 2y$, $y(0) = 3$', fontsize=13)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.xlim(0, 2)
plt.ylim(0, y[-1]*1.05)
plt.tight_layout()
plt.show()

# Print summary
print(f"Solution: y(t) = {y0}*exp({k}*t)")
print(f"Population starts at {y0} and grows at {k*100:.0f}% per time unit.")
print(f"Doubling time ≈ {T_double:.3f} time units")
print(f"y(1) = {y0*np.exp(k*1):.1f},  y(2) = {y0*np.exp(k*2):.1f}")
```

#### Example 1.4: Radioactive Decay  

A radioactive substance decays with decay constant $k = 0.1~\text{year}^{-1}$.  Initial amount: $A_0 = 100$ grams.   Find the amount remaining after **50 years**, and compute the half-life.

Radioactive decay follows:
$
\frac{dA}{dt} = -kA, \quad A(0) = A_0.
$

- **Solution**: $A(t) = A_0 e^{-kt}$  
- **Half-life**: $T_{1/2} = \dfrac{\ln 2}{k}$
  
Given:
- $k = 0.1~\text{yr}^{-1}$  
- $A_0 = 100~\text{g}$  
- $t = 50~\text{yr}$  

Then:
$
A(50) = 100 \cdot e^{-0.1 \times 50} = 100 \cdot e^{-5} \approx 100 \cdot 0.0067379 \approx 0.674~\text{g}.
$

Half-life:$
T_{1/2} = \frac{\ln 2}{0.1} \approx 6.931~\text{years}.$

```python
import numpy as np
import matplotlib.pyplot as plt
# Parameters
A0 = 100.0       # grams
k = 0.1          # per year
t_final = 50     # years

# Time vector
t = np.linspace(0, t_final, 500)
# Analytical solution
A = A0 * np.exp(-k * t)
# Key quantities
A_50 = A0 * np.exp(-k * 50)
T_half = np.log(2) / k
A_at_half = A0 / 2

# Plot
plt.figure(figsize=(8, 5))
plt.plot(t, A, 'b-', linewidth=2.5, label=rf'$A(t) = {A0}e^{{-{k}t}}$')

# Mark t = 50 and half-life
plt.scatter([50], [A_50], color='red', zorder=5, s=80,
            label=rf'$A(50) \approx {A_50:.3f}\,\mathrm{{g}}$')
plt.scatter([T_half], [A_at_half], color='black', zorder=5, s=60,
            label=rf'Half-life $T_{{1/2}} \approx {T_half:.2f}\,\mathrm{{yr}}$')

# Annotations
plt.axvline(T_half, color='gray', linestyle='--', alpha=0.6)
plt.axhline(A_at_half, color='gray', linestyle='--', alpha=0.6)

plt.xlabel('Time $t$ (years)', fontsize=12)
plt.ylabel('Amount $A(t)$ (grams)', fontsize=12)
plt.title('Radioactive Decay ($k = 0.1 {yr}^{-1}$)', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.xlim(0, t_final)
plt.ylim(0, A0 * 1.05)
plt.tight_layout()
plt.show()

# Print results
print(f"Decay constant k = {k:.3f} yr⁻¹")
print(f"Initial amount A₀ = {A0:.1f} g")
print(f"Half-life T_½ = {T_half:.3f} years")
print(f"Amount after 50 years: A(50) = {A_50:.4f} g")
```

### Elementary Theory of Initial-Value Problems

In this chapter, we study the numerical approximation of the initial value problem (IVP) for ordinary differential equations. We begin with the IVP for a first-order ODE:
$$
\left\{ 
\begin{array}{ll}
\frac{dy}{dt} = f(t,y), & a \le t \le b, \\
y(a) = \alpha,
\end{array}
\right.	  	
\label{Eqn:2}
$$ 
where $\alpha$ is a given constant.


 **Definition**  
 A function $f(t, y)$ is said to satisfy a **Lipschitz condition** in the variable $y$ on a set $D \subset \mathbb{R}^2$ if there exists a constant $L > 0$ such that
 $$
 |f(t, y_{1}) - f(t, y_2)| \le L |y_1 - y_2|,
 $$
 whenever $(t, y_1)$ and $(t, y_2)$ are in $D$. The constant $L$ is called a **Lipschitz constant** for $f$.

 **Example**  
 Show that $f(t, y) = 3ty$ satisfies the Lipschitz condition on the region  
 $$
 D = \{(t, y) \mid 1 \le t \le 2,\; -3 \le y \le 4\}.
 $$

**Solution:**  
For any points $(t, y_1)$ and $(t, y_2)$ in $D$, we have  
$ 
|f(t, y_1) - f(t, y_2)| = |3t y_1 - 3t y_2| 
= 3t |y_1 - y_2| 
\le 3 \cdot 2 \cdot |y_1 - y_2| = 6 |y_1 - y_2 | \quad (\text{since } t \le 2).$  
Hence, $f$ satisfies the Lipschitz condition on $D$ with Lipschitz constant $L = 6$.

 **Definition [Convex Set]**  
 A set $D \subset \mathbb{R}^2$ is said to be **convex** if, whenever $(t_1, y_1)$ and $(t_2, y_2)$ belong to $D$, the entire line segment joining them lies in $D$. That is,  
 $$
 (1 - \lambda)(t_1, y_1) + \lambda (t_2, y_2) = \big( (1 - \lambda)t_1 + \lambda t_2,\; (1 - \lambda)y_1 + \lambda y_2 \big) \in D, \;\;
 \text{for all}\;\; \lambda \in [0, 1].
$$

This means that if you pick any two points inside the set, the straight line connecting them stays entirely within the set.

 **Theorem [Sufficient Condition for Lipschitz Condition]**  
 Suppose $f(t, y)$ is defined on a convex set $D \subset \mathbb{R}^2$. If there exists a constant $L$ such that $ \left| \frac{\partial f(t,y)}{\partial y} \right| \le L  \;\text{for all } (t,y) \in D,
 $ then $f$ satisfies a Lipschitz condition in the variable $y$ on $D$ with Lipschitz constant $L$.

### Existence and Uniqueness of Solutions

 **Theorem [Existence and Uniqueness of Solution of First-Order IVPs]**
Suppose $D = \{(t, y) \mid a \le t \le b,\; -\infty < y < \infty\}$ and that $f(t, y)$ is continuous on $D$. If $f$ satisfies a Lipschitz condition in the variable $y$ on $D$, then the initial value problem  
$$
\left\{
\begin{aligned}
\frac{dy}{dt} &= f(t,y), \quad a \le t \le b, \\
y(a) &= \alpha,
\end{aligned}
\right.
$$
 has a unique solution $y(t)$ for all $t \in [a, b]$.

 **Example**  
 Show that there exists a unique solution to the IVP  
 $$
 \left\{
 \begin{aligned}
\frac{dy}{dt} &= 1 + t\sin(ty), \quad 0 \le t \le 2, \\
 y(0) &= 2.
\end{aligned}
 \right.
 $$

**Solution:** In this example, $f(t, y) = 1 + t\sin(ty)$. We compute the partial derivative with respect to $y$:  
$$
\left| \frac{\partial f(t,y)}{\partial y} \right| = \left| t \cdot t\cos(ty) \right| = |t^2 \cos(ty)| \le t^2 \cdot |\cos(ty)| \le 2^2 \cdot 1 = 4,
$$  
since $t \in [0, 2]$, so $t^2 \le 4$, and $|\cos(ty)| \le 1$.  Thus, $\left| \frac{\partial f}{\partial y} \right| \le 4$ on the domain $[0, 2] \times \mathbb{R}$, which implies that $f$ satisfies a Lipschitz condition in $y$ with Lipschitz constant $L = 4$.  
By Theorem [Existence and Uniqueness of Solution of First-Order IVPs], the IVP has a unique solution on $[0, 2]$.

***Well-Posed Problems***

Well-posedness: A mathematical problem is said to be well-posed if it satisfies the following three properties:
  - A solution exists,
  - The solution is unique,
  - The solution depends continuously on the data (e.g., initial conditions and forcing terms), meaning that small changes in the input lead to small changes in the solution.


***Definition*** 
The initial-value problem
$$
\frac{dy}{dt} = f(t, y), \quad a \leq t \leq b, \quad y(a) = \alpha,
$$
is said to be a well-posed problem if:
 - A unique solution $y(t)$ exists on $[a,b]$,
 - There exist constants $\varepsilon_0 > 0$ and $k > 0$ such that for any $\varepsilon$ with $0 < \varepsilon < \varepsilon_0$, whenever $\delta(t)$ is continuous with $|\delta(t)| < \varepsilon$ for all $t \in [a,b]$, and $|\delta_0| < \varepsilon$, the perturbed initial-value problem
   $$
    \frac{dz}{dt} = f(t,z) + \delta(t), \quad a \leq t \leq b, \quad z(a) = \alpha + \delta_0,
    $$
    has a unique solution $z(t)$ satisfying
   $$
    |z(t) - y(t)| < k\varepsilon \quad \text{for all } t \in [a,b].
   $$


The problem perturbed_ivp is called the perturbed problem associated with the original problem. It models the effect of small errors in both the differential equation (via $\delta(t)$) and the initial condition (via $\delta_0$).



Numerical methods will always be concerned with solving a perturbed problem because any round-off error introduced in the representation perturbs the original problem. Unless the original problem is well-posed, there is little reason to expect that the numerical solution to a perturbed problem will accurately approximate the solution to the original problem.
- Now that we know when an IVP has a solution and when it is unique, we can move to the third question: Does the solution depend continuously on the data?
- The well-posedness is important because IVPs obtained from physical observations generally only approximate the true situation, so we need to know whether small errors in the problem statement introduce correspondingly small changes in the solution.
- Another reason is round-off error when numerical methods are used.

***Theorem 1.3[Well-Posedness Theorem]:***
Suppose $D=\{(t,y) \mid a \le t \le b,\; -\infty<y<\infty\}$ and that $f(t, y)$ is continuous and satisfies a Lipschitz condition in $y$ on $D$. Then the initial value problem
$$
\left\{
\begin{aligned}
\frac{dy}{dt} &= f(t,y), \quad a \le t \le b, \\
y(a) &= \alpha,
\end{aligned}
\right.
$$
is well-posed.

***Example:*** Show that the initial value problem
$y' = y - t^2 + 1, \quad 0 \le t \le 2, \quad y(0) = 0.5$
is well-posed on $D = \{(t, y) \mid 0 \le t \le 2,\; -\infty < y < \infty\}$.

***Solution:***  Here, $f(t,y) = y - t^2 + 1$. This function is continuous on $D$, and
$$
\left| \frac{\partial f}{\partial y} \right| = \left| \frac{\partial}{\partial y}(y - t^2 + 1) \right| = 1.
$$
Since the partial derivative is bounded, $f$ satisfies a Lipschitz condition in $y$ with Lipschitz constant $L = 1$. By Theorem 1.3, the IVP is well-posed.

As an illustration, consider the perturbed problem:
$$
\frac{dz}{dt} = z - t^2 + 1 + \delta(t), \quad 0 \leq t \leq 2, \quad z(0) = 0.5 + \delta_0,
$$
where $\delta(t)$ is a continuous function and $\delta_0$ is a small constant. Suppose $|\delta(t)| < \varepsilon$ and $|\delta_0| < \varepsilon$ for some $\varepsilon > 0$.
The exact solution to the original IVP is:
$y(t) = (t + 1)^2 - 0.5 e^t.$

The solution to the perturbed problem (with constant $\delta(t) = \delta$) is:
$$
z(t) = (t + 1)^2 + (\delta + \delta_0 - 0.5)e^t - \delta.
$$
Then, for $t \in [0, 2]$,
$$
\begin{align*}
|y(t) - z(t)| &= \left| \big[(t+1)^2 - 0.5 e^t\big] - \big[(t+1)^2 + (\delta + \delta_0 - 0.5)e^t - \delta\big] \right| 
&= \left| -0.5 e^t - (\delta + \delta_0 - 0.5)e^t + \delta \right| \\
&= \left| -(\delta + \delta_0)e^t + \delta \right| \\
&\leq |\delta + \delta_0| e^t + |\delta| \\
&\leq (|\delta| + |\delta_0|) e^2 + |\delta| < (\varepsilon + \varepsilon) e^2 + \varepsilon = (2e^2 + 1)\varepsilon.
\end{align*} $$
Thus, $|z(t) - y(t)| < k \varepsilon$ with $k = 2e^2 + 1$, independent of $t$. This confirms that the problem is well-posed.

## 1.2 Introduction to BVPs

While initial value problems (IVPs) model time evolution with data specified at a single point, boundary value problems (BVPs) describe equilibrium states where conditions are imposed at two distinct points—typically the endpoints of an interval $[a,b]$.

A typical second-order BVP has the form:
$$
  \begin{equation}
    y''(t) = f(t, y(t), y'(t)), \quad a < t < b,
\end{equation}
$$
subject to boundary conditions such as:

   -  Dirichlet: $ y(a) = \alpha, \; y(b) = \beta $
   -   Neumann: $ y'(a) = \gamma, \; y'(b) = \delta $
   -    Robin (mixed): $ \alpha_1 y(a) + \alpha_2 y'(a) = \gamma, \; \beta_1 y(b) + \beta_2 y'(b) = \delta $

Unlike IVPs, where existence and uniqueness are typically guaranteed under mild conditions, BVPs exhibit more complex behavior: solutions may not exist, may be **non-unique**, or may depend sensitively on boundary data.

 Physical problems that are position-dependent rather than time-dependent are often described by differential equations with conditions imposed at more than one point.
    

  
The two-point boundary-value problems in this course involve a second-order differential equation of the form:
$$
\begin{equation}
y'' = f(x, y, y'), \quad a < x < b
\label{eq:bvp-ode}
\end{equation}
$$
together with the boundary conditions
$$
\begin{equation}
y(a) = \alpha, \quad y(b) = \beta.
\label{eq:bvp-bc}
\end{equation}
$$
 
 The following theorem gives general conditions that ensure the existence and uniqueness of the solution to a second-order boundary value problem. The proof can be found in \cite{Keller}.


### 1.2.1 Existence and Uniqueness of Solutions of BVPs	
### Theorem 1  <a id="th1"></a>  
Suppose the function $ f $ in the boundary value problem 

$ y'' = f(x, y, y'),    a < x < b, $ 

$y(a) = \alpha,   y(b) = \beta, $

is continuous on the domain  $D = \left\{ (x, y, y') \mid a < x < b,\; -\infty < y < \infty,\; -\infty < y' < \infty \right\},
$  
and that the partial derivatives $ f_y $ and $ f_{y'} $ are also continuous on $ D $.  

If the following conditions hold:

1. $ f_y(x, y, y') > 0 $ for all $ (x, y, y') \in D $,  
2. There exists a constant $ M > 0 $ such that  
   $$
   |f_{y'}(x, y, y')| \leq M \quad \text{for all } (x, y, y') \in D,
   $$

then the boundary-value problem has a **unique solution** on $[a, b]$.

### Example:  
Use Theorem 1 to show that the boundary-value problem
$ y'' + e^{-x y} + \sin(y') = 0, \qquad 1 < x < 2, \qquad y(1) = y(2) = 0$
has a unique solution.

### Solution: 
Rewrite the differential equation in the standard form
$
y'' = f(x, y, y'),$ 
where $
f(x, y, y') = -e^{-x y} - \sin(y').$

To apply Theorem 1, we verify its two key hypotheses on the domain $[1,2] \times \mathbb{R} \times \mathbb{R}$:

1. **Monotonicity in $ y $:**  
   The partial derivative of $ f $ with respect to $ y $ is  
   $$
   f_y(x, y, y') = \frac{\partial}{\partial y}\bigl(-e^{-x y} - \sin(y')\bigr) = x e^{-x y}.
   $$  
   Since $ x \in [1, 2] $ implies $ x > 0 $, and the exponential function satisfies $ e^{-x y} > 0 $ for all real $ y $, it follows that  
   $$
   f_y(x, y, y') = x e^{-x y} > 0 \quad \text{for all } (x, y, y') \in D.
   $$


2. **Boundedness of the derivative with respect to y'**  
   The partial derivative of $ f $ with respect to $ y' $ is  
   $$
   f_{y'}(x, y, y') = \frac{\partial}{\partial y'}\bigl(-e^{-x y} - \sin(y')\bigr) = -\cos(y').
   $$  
   Since $ |\cos(y')| \leq 1 $ for all real $ y' $, we obtain the uniform bound  
   $$
   |f_{y'}(x, y, y')| = |\cos(y')| \leq 1 \quad \text{for all } (x, y, y') \in D.
   $$

Both hypotheses of [Theorem 1](#th1) are satisfied on the domain $ D = \{(x, y, y') \mid x \in [1,2],\, y, y' \in \mathbb{R} \} $.  
 Therefore, the boundary-value problem  
$$
y'' = -e^{-x y} - \sin(y'), \quad y(1) = \alpha,\; y(2) = \beta
$$  
admits a **unique solution** on $[1, 2]$.

$\blacksquare$

### For Linear Boundary-Value Problems

The second-order differential equation  
$$
y'' = f(x, y, y')
$$  
is said to be **linear** if the function $ f $ can be written in the form  
$$
f(x, y, y') = p(x)\,y' + q(x)\,y + r(x),
$$  
where $ p(x) $, $ q(x) $, and $ r(x) $ are given functions depending *only* on $ x $.

Linear boundary-value problems of this type arise frequently in physics, engineering, and applied mathematics (e.g., beam deflection, steady-state heat conduction, quantum mechanics).

In this special case, the general uniqueness result stated in [Theorem 1](#th1) simplifies significantly. Substituting  
$$
f_y = q(x), \qquad f_{y'} = p(x),
$$  
the hypotheses become:
- $ q(x) > 0 $ on $[a, b]$ (monotonicity in $ y $),  
- $ |p(x)| \le M $ for some constant $ M $ (bounded damping/convection coefficient).

Hence, uniqueness follows directly under these elementary conditions on the coefficient functions.

### Theorem 2  <a id="th2"></a>  
Consider the linear boundary-value problem  
$$
y'' = p(x)\,y' + q(x)\,y + r(x), \qquad a < x < b, \label{eq3} \tag{3}
$$  
subject to the boundary conditions  
$$
y(a) = \alpha, \qquad y(b) = \beta. \label{eq4} \tag{4}
$$  

Assume the following:

1. The coefficient functions $ p(x) $, $ q(x) $, and $ r(x) $ are **continuous** on the closed interval $[a, b]$;  
2. The coefficient $ q(x) $ is **strictly positive** on $[a, b]$, i.e.,  
   $$
   q(x) > 0 \quad \text{for all } x \in [a, b].
   $$

Then the boundary-value problem \eqref{eq3}–\eqref{eq4} admits a **unique solution** on $[a, b]$.

### Examples Illustrating Theorem 2

#### Example 1  
Consider the BVP:  
$$
y'' = 2y' + 3y + \sin x, \qquad 0 < x < \pi,
$$  
with boundary conditions  
$$
y(0) = 0, \quad y(\pi) = 1.
$$

- $ p(x) = 2 $, $ q(x) = 3 $, $ r(x) = \sin x $.  
- All coefficients are continuous on $[0, \pi]$.  
- $ q(x) = 3 > 0 $ for all $ x \in [0, \pi] $.

**Conclusion**: Hypotheses of [Theorem 2](#th2) are satisfied ⇒ **unique solution exists**.

---




#### Example 2  
BVP:  
$$
y'' = x\,y' + (1 + x^2)\,y + e^x, \qquad -1 < x < 1,
$$  
with  
$$
y(-1) = 2, \quad y(1) = -1.
$$

- $ p(x) = x $, $ q(x) = 1 + x^2 $, $ r(x) = e^x $.  
- All continuous on $[-1, 1]$.  
- Since $ x^2 \ge 0 $, we have $ q(x) = 1 + x^2 \ge 1 > 0 $ on $[-1, 1]$.

 **Conclusion**: Conditions of [Theorem 2](#th2) hold ⇒ **unique solution exists**.



####  Example 3  
BVP:  
$$
y'' = \cos(x)\,y' + \bigl(2 + \sin x\bigr)\,y + x^2, \qquad 0 < x < \tfrac{\pi}{2},
$$  
with  
$$
y(0) = 1, \quad y\!\left(\tfrac{\pi}{2}\right) = 0.
$$

- $ p(x) = \cos x $, $ q(x) = 2 + \sin x $, $ r(x) = x^2 $.  
- All continuous on $\left[0, \tfrac{\pi}{2}\right]$.  
- Since $ \sin x \in [0, 1] $ on this interval, $ q(x) \in [2, 3] $, so $ q(x) \ge 2 > 0 $.

 **Conclusion**: Satisfies [Theorem 2](#th2) ⇒ **unique solution exists**.

####  Example 4: Failure of Positivity → Non-uniqueness  
BVP:  
$$
y'' = -\pi^2 y, \qquad 0 < x < 1,
$$  
with homogeneous boundary conditions  
$$
y(0) = 0, \quad y(1) = 0.
$$

- $ p(x) = 0 $, $ q(x) = -\pi^2 < 0 $, $ r(x) = 0 $.  
- Coefficients are continuous, but **$ q(x) \not> 0 $** — violates Theorem 2.

Indeed, every function of the form  
$$
y(x) = C \sin(\pi x), \quad C \in \mathbb{R},
$$  
satisfies the ODE and both boundary conditions.

**Infinitely many solutions** ⇒ **uniqueness fails**.

The strict positivity of $ q(x) $ in [Theorem 2](#th2) is **essential**.

## 1.3 Introduction to Partial Differential Equations (PDEs)

A **partial differential equation (PDE)** is a differential equation that involves **partial derivatives** of one or more dependent variables with respect to **two or more independent variables**.

PDEs are the cornerstone of mathematical modeling in science and engineering. They arise naturally when describing continuous systems where change occurs in multiple directions, such as:

-  **Heat conduction** (e.g., the heat equation: $ u_t = \alpha \nabla^2 u $)  
-  **Wave propagation** (e.g., the wave equation: $ u_{tt} = c^2 \nabla^2 u $)  
-  **Fluid dynamics** (e.g., Navier–Stokes equations)  
-  **Electrostatics & electromagnetism** (e.g., Laplace’s equation $ \nabla^2 \phi = 0 $, Maxwell’s equations)  
-  **Quantum mechanics** (e.g., Schrödinger equation)

Unlike ordinary differential equations (ODEs), whose solutions depend on one independent variable (e.g., time), PDE solutions depend on *multiple* variables (e.g., space *and* time), making their analysis — both analytical and numerical — significantly richer and more challenging.

### Homogeneous vs. Non-Homogeneous PDEs

A PDE is **homogeneous** if *every term* contains the dependent variable $ u $ or one of its partial derivatives.  
Otherwise, it is **non-homogeneous** (i.e., it contains a source or forcing term independent of $ u $).

 From the earlier list:
- **Homogeneous**:  
  (1) $ u_{xx} + u_{yy} = 0 $  
  (3) $ u_{xx} - u_{yy} = 0 $  
  (4) $ a^2 u_{xx} - u_{tt} = 0 $  
  (5) $ u_{xx} + (1 + u_x^2)u_{yy} = 0 $  
  (6) $ x u_{xx} + 2xy\,u\,u_{xy} + y u_{yy} = 0 $

- **Non-homogeneous**:  
  (2) $ u_{xx} + u_{yy} = f(x, y) $  
  → The term $ f(x, y) $ is independent of $ u $, acting as a source.

> 💡 Note: Homogeneity is *independent* of linearity. For example, (5) and (6) are nonlinear *and* homogeneous.

---

### Notation for Partial Derivatives

We use subscript notation:
$$
u_x = \frac{\partial u}{\partial x}, \quad
u_{xx} = \frac{\partial^2 u}{\partial x^2}, \quad
u_{xy} = \frac{\partial^2 u}{\partial x\,\partial y}, \quad \text{etc.}
$$

**Examples in 3D**:
- **Laplace equation**:  
  $$
  u_{xx} + u_{yy} + u_{zz} = 0 \tag{7}
  $$
- **Poisson equation**:  
  $$
  u_{xx} + u_{yy} + u_{zz} = f(x, y, z) \tag{8}
  $$

---

### Classification of Second-Order Linear PDEs (Two Variables)

Consider the general second-order PDE (with coefficients depending only on $ x, y $ — i.e., *linear* or *semi-linear* principal part):
$$
A(x,y)\,u_{xx} + B(x,y)\,u_{xy} + C(x,y)\,u_{yy} + D u_x + E u_y + F u = G \tag{9}
$$

The **type** of the PDE at a point is determined by the **discriminant**:
$$
\Delta(x, y) = B^2 - 4AC
$$

| Type        | Discriminant      | Physical Example        | Canonical Form         |
|-------------|-------------------|--------------------------|------------------------|
| **Elliptic**   | $ \Delta < 0 $  | Steady-state heat flow, electrostatics | $ u_{xx} + u_{yy} = 0 $ |
| **Parabolic**  | $ \Delta = 0 $  | Heat/diffusion equation  | $ u_t - \alpha u_{xx} = 0 $ |
| **Hyperbolic** | $ \Delta > 0 $  | Wave propagation         | $ u_{tt} - c^2 u_{xx} = 0 $ |

###  Applying the classification to our examples:

| Eq. | PDE | $ A $ | $ B $ | $ C $ | $ \Delta = B^2 - 4AC $ | Type |
|-----|-----|--------|--------|--------|--------------------------|------|
| (1) | $ u_{xx} + u_{yy} = 0 $ | 1 | 0 | 1 | $ 0 - 4(1)(1) = -4 < 0 $ | **Elliptic** |
| (2) | $ u_{xx} + u_{yy} = f $ | 1 | 0 | 1 | $ -4 < 0 $ | **Elliptic** |
| (3) | $ u_{xx} - u_{yy} = 0 $ | 1 | 0 | -1 | $ 0 - 4(1)(-1) = 4 > 0 $ | **Hyperbolic** |
| (4) | $ a^2 u_{xx} - u_{tt} = 0 $ | $ a^2 $ | 0 | $-1$ | $ 0 - 4(a^2)(-1) = 4a^2 > 0 $ | **Hyperbolic** |

>  **Important**: This discriminant classification applies *only* when $ A, B, C $ depend on $ x, y $ alone (i.e., for linear or semi-linear PDEs). For quasi-linear PDEs like (5) and (6), the type may *vary with the solution* (e.g., equation (5) is elliptic where $ 1 + u_x^2 > 0 $, which is always true — so it remains elliptic everywhere).


### Examples of Classification by Discriminant

Recall: For a second-order linear PDE  
$$
A u_{xx} + B u_{xy} + C u_{yy} + \text{(lower-order terms)} = 0,
$$  
the **discriminant** is $ \Delta = B^2 - 4AC $, and:
- $ \Delta < 0 $ → **Elliptic**
- $ \Delta = 0 $ → **Parabolic**
- $ \Delta > 0 $ → **Hyperbolic**

---

### 1️⃣ Laplace Equation — *Elliptic*  
**PDE**:  
$$
u_{xx} + u_{yy} = 0
$$  
- $ A = 1 $, $ B = 0 $, $ C = 1 $  
- Discriminant:  
  $$
  \Delta = 0^2 - 4(1)(1) = -4 < 0
  $$  
 **Elliptic** — models steady-state phenomena (e.g., electrostatic potential, incompressible flow).

---

### 2️⃣ Heat Equation — *Parabolic*  
**PDE** (1D, $ t $ = time):  
$$
u_t = k u_{xx} \quad \Longleftrightarrow \quad k u_{xx} - u_t = 0
$$  
Treat as PDE in $ (x, t) $: only second derivative is $ u_{xx} $, so  
- $ A = k $, $ B = 0 $, $ C = 0 $  
- Discriminant:  
  $$
  \Delta = 0^2 - 4(k)(0) = 0
  $$  
 **Parabolic** — models diffusion processes (irreversible time evolution).

---

### 3️⃣ Wave Equation — *Hyperbolic*  
**PDE** (1D):  
$$
u_{tt} = c^2 u_{xx} \quad \Longleftrightarrow \quad c^2 u_{xx} - u_{tt} = 0
$$  
In variables $ (x, t) $:  
- $ A = c^2 $, $ B = 0 $, $ C = -1 $  
- Discriminant:  
  $$
  \Delta = 0^2 - 4(c^2)(-1) = 4c^2 > 0 \quad (c \ne 0)
  $$  
 **Hyperbolic** — models wave propagation, finite signal speed, reversibility.
